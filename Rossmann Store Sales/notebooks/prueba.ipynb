# Model.py

```python
import pandas as pd

df = pd.read_parquet("../data/processed/train_features.parquet")
df.head()
```

```python
val_df = df[(df["Year"] == 2015) & (df["Month"] >= 2)]
```

```python
train_df = df[(df["Year"] != 2015) | (df["Month"] == 1) & (df["Year"] == 2015)]
```

```python
df.shape[0] == train_df.shape[0] + val_df.shape[0]
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GridSearchCV

features = ['Store', 'Customers', 'Open', 'Promo', 'SchoolHoliday', 'Year', 'Month', 'Day', 'Sales_lag_1', 'Sales_lag_7', 'rolling_mean_7']
X_train = train_df[features]
y_train = train_df['Sales']
X_val = val_df[features]
y_val = val_df['Sales']

rf = RandomForestRegressor(
    n_estimators=200,          # número de árboles (200 suele ser buen equilibrio)
    max_depth=15,              # profundidad máxima de cada árbol
    min_samples_split=5,       # mínimo de muestras para dividir un nodo
    min_samples_leaf=2,        # mínimo de muestras en una hoja
    max_features='sqrt',       # número de features a considerar por división
    n_jobs=-1,                 # usa todos los núcleos del procesador
    random_state=42,           # reproducibilidad
    bootstrap=True             # con reemplazo (por defecto)
)

param_grid = {
    'n_estimators': [100, 200, 500],
    'max_depth': [10, 15, 20, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'max_features': ['sqrt', 'log2', None],
}

grid = GridSearchCV(rf, param_grid, cv=3, scoring="neg_mean_absolute_error", n_jobs=-1)
grid.fit(X_train, y_train)

print(grid.best_params_)

#y_pred = rf.predict(X_val)
```

```python
from tqdm.notebook import tqdm
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GridSearchCV

# tqdm se integra automáticamente con GridSearchCV si se usa con 'verbose'
# Pero para una barra de progreso visual, puedes usar tqdm manualmente en un ciclo, por ejemplo:


features = ['Store', 'Customers', 'Open', 'Promo', 'SchoolHoliday', 'Year', 'Month', 'Day', 'Sales_lag_1', 'Sales_lag_7', 'rolling_mean_7']
X_train = train_df[features]
y_train = train_df['Sales']
X_val = val_df[features]
y_val = val_df['Sales']

rf = RandomForestRegressor(
    n_estimators=200,
    max_depth=15,
    min_samples_split=5,
    min_samples_leaf=2,
    max_features='sqrt',
    n_jobs=-1,
    random_state=42,
    bootstrap=True
)

param_grid = {
    'n_estimators': [100, 200, 500],
    'max_depth': [10, 15, 20, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'max_features': ['sqrt', 'log2', None],
}

# Calcula el número total de combinaciones
total = 1
for v in param_grid.values():
    total *= len(v)

with tqdm(total=total, desc="GridSearchCV Progress") as pbar:
    class TQDMGridSearchCV(GridSearchCV):
        def _run_search(self, evaluate_candidates):
            def wrapped(candidate_params):
                pbar.update(len(candidate_params))
                return evaluate_candidates(candidate_params)
            super()._run_search(wrapped)

    grid = TQDMGridSearchCV(rf, param_grid, cv=3, scoring="neg_mean_absolute_error", n_jobs=-1)
    grid.fit(X_train, y_train)

print(grid.best_params_)
```
