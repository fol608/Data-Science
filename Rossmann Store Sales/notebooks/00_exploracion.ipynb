```python
import matplotlib.pyplot as plt
import math

columns_hist_list = list(df.columns)
columns_hist_list.remove("Date")
columns_hist_list.remove("StateHoliday")

size = len(columns_hist_list)
rows_hist = math.ceil(size / 4)

plt.figure(figsize=(16, 4 * rows_hist))
for idx, c in enumerate(columns_hist_list):
    plt.subplot(rows_hist, 4, idx + 1)
    plt.hist(df[c])
    plt.title(c)

plt.tight_layout()
plt.show()

```
